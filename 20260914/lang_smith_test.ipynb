```python
#!pip install -qU langchain "langchain[openai]"
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain.agents import create_agent

@tool
def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

#2. 모델, 에이전트 연결
llm = ChatOpenAI(model="gpt-4o-mini",max_completion_tokens=2048)
agent = create_agent(model=llm, tools=[get_weather])

result = agent.invoke({"messages": [("user","San Francisco 날씨 어때?")]})
print(result["messages"][-1].content)

```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini",
                 temperature=0.1,
                 max_completion_tokens=2048,
                 ).bind(logprobs=True)
question = "양자화와 양자프로그래밍에 대해 설명해줘."
response = llm.invoke(question)
```

```python
response.response_metadata
```

```python
print(response.content)
```

```python
answer = llm.stream("그로버 양자 알고리즘에 대해서 설명해줘.")
final_answer = "" 

for token in answer:
    final_answer+=token.content
    print(token.content,end="",flush=True)
```

```python
print(final_answer)
```

```python
import logging
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_teddynote.messages import stream_response
```

```python
#python logging 파일 설정(app.log 파일로 설정)
logging.basicConfig(
    filename="app.log",                                   # 저장될 로그 파일 이름
    level=logging.INFO,                                   # 로깅 레벨(INFO, DEBUG, ERROR 등)
    format="%(asctime)s - %(levelname)s - %(message)s",   # 로그 포맷 (시간 - 레벨 - 메시지)
    encoding="utf-8",                                     # 한글 깨짐 방지
)

#로컬 로깅 경로 
logger = logging.getLogger("MyMultimodalApp")

#로깅 시작 기록
logger.info("LangSmith 연동 및 애플리케이션 시작")

```

```python
try:
    multimodal_llm = ChatOpenAI(model = "gpt-4o-mini", temperature=0)
    IMAGE_URL = "https://t3.ftcdn.net/jpg/03/77/33/96/360_F_377339633_Rtv9I77sSmSNcev8bEcnVxTHrXB4nRJ5.jpg"
    prompt = "이미지의 내용을 한국어로 상세히 설명해 주세요."
    
    logger.info(f"요청 전송 - Prompt: {prompt} / Image: {IMAGE_URL}")
    
    # 메시지 생성 및 멀티모달 호출
    message = HumanMessage(
        content=[
            {"type": "text", "text": prompt},
            {"type": "image_url", "image_url": {"url": IMAGE_URL}},
        ]
    )
    
    #스트리밍 및 응답 기록
    answer = multimodal_llm.stream([message])
    
    logger.info("답변 수신 및 스트리밍 시작")
    final_answer=""
    for token in answer:
        final_answer+=token.content
        print(token.content)
        
    logger.info("답변 처리 완료 ")
    
except Exception as e:
    # 에러 발생 시 로그 파일에 에러 기록
    logger.error(f"실행 중 에러 발생: {str(e)}", exc_info=True)
    
```
