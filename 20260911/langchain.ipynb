```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm_gpt_with_logprob = ChatOpenAI(model="gpt-4o-mini",temperature=0.1,max_completion_tokens=2048,).bind(logprobs=True)
```

```python
question = "대한민국의 수도는 어디인가요?"
response = llm_gpt_with_logprob.invoke(question)
response.response_metadata
```
